# Polynomial Regression vs Linear Regression
Comparing Linear and Polynomial (degree 4) regression models on position-salary data.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 1. Load and Inspect Dataset

```python
dataset = pd.read_csv('position_salaries.csv')
dataset.info()
dataset.head()
```

## 2. Prepare Features and Target

```python
X = dataset.iloc[:, 1:2].values  # Position level (2D array)
y = dataset.iloc[:, 2].values    # Salary
```

## 3. Train/Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0
)
```

## 4. Linear Regression

```python
from sklearn.linear_model import LinearRegression

lin_reg = LinearRegression()
lin_reg.fit(X, y)
print(lin_reg)
```

```python
def viz_linear():
    plt.scatter(X, y, color='red', label='Actual')
    plt.plot(X, lin_reg.predict(X), color='blue', label='Predicted')
    plt.title('Linear Regression Model')
    plt.xlabel('Position Level')
    plt.ylabel('Salary')
    plt.legend()
    plt.tight_layout()
    plt.show()

viz_linear()
```

## 5. Polynomial Regression (Degree 4)

```python
from sklearn.preprocessing import PolynomialFeatures

poly_reg = PolynomialFeatures(degree=4)
X_poly = poly_reg.fit_transform(X)

pol_reg = LinearRegression()
pol_reg.fit(X_poly, y)
print(pol_reg)
```

```python
def viz_polynomial():
    X_grid = np.arange(min(X), max(X), 0.1).reshape(-1, 1)  # Smoother curve
    plt.scatter(X, y, color='red', label='Actual')
    plt.plot(X_grid, pol_reg.predict(poly_reg.fit_transform(X_grid)), color='blue', label='Predicted')
    plt.title('Polynomial Regression (Degree 4)')
    plt.xlabel('Position Level')
    plt.ylabel('Salary')
    plt.legend()
    plt.tight_layout()
    plt.show()

viz_polynomial()
```

## 6. Predictions for Level 5.5

```python
level = [[5.5]]

linear_pred = lin_reg.predict(level)[0]
poly_pred = pol_reg.predict(poly_reg.fit_transform(level))[0]

print(f'Linear Regression Prediction for Level 5.5:     ${linear_pred:,.2f}')
print(f'Polynomial Regression Prediction for Level 5.5: ${poly_pred:,.2f}')
```
